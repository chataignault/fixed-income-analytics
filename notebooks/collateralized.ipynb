# Collateralized trade, cash flows and risk

- no risk-free rate,
- price becomes the process that defines level of collateral holding,
- futures contracts are a particular type of collateralized forward contract (with collateral rate set to zero),
- $\text{MTM} + \text{Collateral} = 0$ for each party,
- a collateralized trade can be terminated at any moment at zero cost,
- a collateralized stock sale is actually a repo transaction,
